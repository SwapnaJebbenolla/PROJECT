```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

product_name = []
price = []
rating = []
availability = []
product_url = []
```

```python
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}
```

```python
base_url = "https://books.toscrape.com/catalogue/page-{}.html"

headers = {
    "User-Agent": "Mozilla/5.0"
}

for page in range(1, 51):

    print(f"Scraping Page {page}")

    url = base_url.format(page)

    response = requests.get(url, headers=headers)

    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    for book in books:

        # Product Name
        title = book.h3.a["title"]

        # Price
        book_price = book.find("p", class_="price_color").text.replace("£", "")

        # Availability
        stock = book.find("p", class_="instock availability").text.strip()

        # Rating
        rating_text = book.find("p")["class"][1]
        book_rating = rating_map[rating_text]

        # Product URL
        link = "https://books.toscrape.com/catalogue/" + book.h3.a["href"].replace("../", "")

        product_name.append(title)
        price.append(book_price)
        availability.append(stock)
        rating.append(book_rating)
        product_url.append(link)

    time.sleep(1)

print("Scraping Completed")
```

```python
df = pd.DataFrame({
    "Product_Name": product_name,
    "Price": price,
    "Availability": availability,
    "Rating": rating,
    "Product_URL": product_url
})
```

```python
import pandas as pd
import numpy as np
import random

# Read your scraped CSV
df = pd.read_csv("capstone project raw_products data.csv")

# Check data
df.head()
```

```python
df.drop(columns=["Product_URL"], inplace=True)
```

```python
df.columns
```

```python
df["Selling_Price"] = (
    df["Selling_Price"]
    .astype(str)
    .str.replace("Â", "", regex=False)
    .str.replace("£", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["Selling_Price"] = pd.to_numeric(df["Selling_Price"], errors="coerce")
```

```python
df.head()
```

```python
df["Discount_Percentage"] = (
    (
        df["Original_Price"] - df["Selling_Price"]
    ) / df["Original_Price"] * 100
).round(2)
```

```python
df[["Selling_Price","Original_Price","Discount_Percentage"]].head()
```

```python
book_sellers = [
    "Book World",
    "Readers Hub",
    "Book Bazaar",
    "Book Planet",
    "Book Corner",
    "Readers Choice",
    "Library Store",
    "Book House",
    "Online Book Shop",
    "Book Galaxy"
]

df["Book_Seller"] = np.random.choice(book_sellers, len(df))
```

```python
df["Delivery_Days"] = np.random.randint(1,8,len(df))
```

```python
df["Free_Delivery"] = np.random.choice(
    ["Yes","No"],
    len(df),
    p=[0.80,0.20]
)
```

```python
df["Premium_Delivery"] = np.random.choice(
    ["Yes","No"],
    len(df),
    p=[0.70,0.30]
)
```

```python
df["Wishlist_Count"] = np.random.randint(0,2500,len(df))
```

```python
df["Copies_Sold"] = np.random.randint(50,8000,len(df))
```

```python
df["Return_Rate"] = np.round(
    np.random.uniform(0,10,len(df)),
    2
)
```

```python
df["Publication_Year"] = np.random.randint(2015,2026,len(df))
```

```python
cities = [
    "Hyderabad",
    "Bengaluru",
    "Chennai",
    "Mumbai",
    "Delhi",
    "Pune",
    "Kolkata",
    "Ahmedabad",
    "Jaipur",
    "Lucknow",
    "Bhubaneswar",
    "Visakhapatnam"
]

df["Warehouse_City"] = np.random.choice(cities, len(df))
```

```python
conditions = [
    df["Customer_Rating"] >= 4,
    df["Customer_Rating"] >= 3
]

choices = [
    "High",
    "Medium"
]

df["Customer_Satisfaction"] = np.select(
    conditions,
    choices,
    default="Low"
)
```
